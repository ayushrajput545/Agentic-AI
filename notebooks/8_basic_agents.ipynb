```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool # import tool decorator
@tool
def add_numbers(a:int , b:int):
    """
      it will return the sum of two numebers
      Args:
        a:Number one
        b:Number Two
    """
    return a+b

@tool
def mul_numbers(a:int , b:int):
    """
      it will return the product of two numebers
      Args:
        a:Number one
        b:Number Two
    """
    return a*b
```

```python
add_numbers.invoke({"a":12 , "b":3})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model='gemini-3.6-flash')
from langchain.agents import create_agent
 

# AGENT = LLM + tools
agent = create_agent(
    model=llm,
    tools=[add_numbers , mul_numbers],
    system_prompt="You are a Math teacher and always use tool for calcualtion"
)

res =agent.invoke({"messages":[{"role":"user" , "content":"What is 2 + 2 *7 ?"}]}) # fix syntax
res
```

Question -> Agent -> Tool call , this tool return some ans -> LLM observe the response return from tool and generate answer
