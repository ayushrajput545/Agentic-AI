```python
from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model='gemini-3.6-flash')
```

#### Discuss the problem first

```python
text = "My name is AYush , I am 22 year old , My email id is ayy@gmail.comm"
res = llm.invoke(f"Give me only name , age , and email from this text {text}")
res.content[0]["text"] # not good structired format of this
```

#### SOlution: Strcutre format

```python
from pydantic import BaseModel , Field

class ResponseStructure(BaseModel):
    name:str = Field(description = "Complete Name")
    email:str = Field(description = "Email")
    age:int = Field(description = "Age of User")

structured_llm = llm.with_structured_output(ResponseStructure)
res = structured_llm.invoke(f"Give me only name , age , and email from this text {text}")
res
```
