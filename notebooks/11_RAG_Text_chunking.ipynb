```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = "Hi , I am Ayush Rajput , a final year Student at IET Lucknow"\
"I am a full stack developer with expertise in javascript and MERN stack"

splitter = RecursiveCharacterTextSplitter(chunk_size=50, chunk_overlap=20) #mean first chunk ke charters next mein overlap bhi kar skte (20 charcter)
chunks = splitter.split_text(text)
print(chunks) #list of chunks
```

```python
#document loader
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
pdfloader = PyPDFLoader('../data/Ayush_resume.pdf')
docs = pdfloader.load()

splitter = RecursiveCharacterTextSplitter(chunk_size=500 , chunk_overlap=20)
chunks = splitter.split_documents(docs)
print(chunks)
len(chunks) #list
```
