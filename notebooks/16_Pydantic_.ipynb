```python

from pydantic import BaseModel, Field

class User(BaseModel):
    title: str = Field(max_length=10) #max 10 characters only
    des: str = Field(max_length=20)
    price:int=Field(gt=0,le=100) #greater 0 and less than 100

userObj = User(title="adjoid", des="def" , price=99)

print(userObj)
```

```python
#we connect one class inside another class
class Order(BaseModel):
    id:str
    user:User

order = Order(id="lol",user=userObj) # wither we pass dict or product itself
order
```

```python
#concept of list , if we want to validate any list
from typing import List
class Cart(BaseModel):
    id:str
    items:List[User] #[User1 , User2] ; List[str] -> string
    comment:str = Field(default="No Comments")

cart = Cart(id="dwsd", items=[userObj])
print(cart)
for user in cart.items:
    print(user.title)
```

```python
#Annoted -> Allow us to build custome validator
from typing import Annotated
commonValidator = Annotated[List[User], "This contain List of users"]

#Now above commonvalidator can be used in above class like this 
class Cart(BaseModel):
    id:str
    items:commonValidator #[User1 , User2] ; List[str] -> string
    comment:str = Field(default="No Comments")
```
