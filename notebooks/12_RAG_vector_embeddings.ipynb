```python
from dotenv import load_dotenv
load_dotenv()
```

```python
document =[
    "I am Ayush Rajput",
    "I have some working knowledge with GEN AI",
    "I Build RAG Systems , AI AGENTS",
    "I also knows the basics of Machine Learning"
]
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
embeddings = GoogleGenerativeAIEmbeddings(model='gemini-embedding-001') #produces 3072-dimensional embeddings by default,
vector = embeddings.embed_query(document)
print(vector)
```

```python
len(vector)
```

```python
from langchain_chroma import Chroma #vector db

#it convert our data into vector embedings also store in vector db
vector_store = Chroma.from_texts(
    texts=document,
    embedding=embeddings,
    persist_directory="./vector_db" #persist this data in this particular folder
)
```

```python
query = 'Machine Learning'
result = vector_store.similarity_search(query=query , k=2) #only two chunks
result
```
