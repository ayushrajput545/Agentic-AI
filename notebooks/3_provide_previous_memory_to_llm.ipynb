```python
from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model='gemini-3.6-flash')
```

```python
res = llm.invoke("Hello! My name is Ayush")
res.content[0]["text"]
```

```python
res = llm.invoke("What is My name ?")
res.content[0]["text"]
```

```python
prompts = [
    {"role":"user" , "content":"My Name is Ayush"},
    {"role":"ai" , "content":"Hey Nice to meet you "},
    {"role":"user", "content":"What is My name?"}
]

res = llm.invoke(prompts)
res.content[0]["text"]
```

```python
# standard way maintain 
history = [] # list of dict [{"role":"user" , "content":"My Name is Ayush"},]
while True:
    query = input("User: ")
    if query.lower() in ["bye" , "stop", "exit"]:
        print("Good Bye!")
        break
    
    history.append({"role":"user" , "content":query})
    print("User:", query)

    res = llm.invoke(history)
    history.append({"role":"ai", "content":res.content })
    print("AI:", res.content[0]["text"])
```
