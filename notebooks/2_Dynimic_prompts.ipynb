```python
## Dynimic Prompts -> No need to wrtei again and again your are java developer , your are javascript developer , python developer

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
 
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash")
```

```python
prompts =ChatPromptTemplate.from_messages([
    {"role":"system" , "content":"You are {language} developer"},
    {"role":"user" , "content":"{query}"}
])
prompts
```

```python
final_prompts = prompts.format_messages(language="Java" , query="Explain Lambda function in 20 words") 
final_prompts
res  = llm.invoke(final_prompts)
res
```

```python
## concepts of chain - runnable
res = llm.invoke(prompts.invoke({  # here llm.invoke and prompts.invoke are runnable both promppmt.invoke ka output llm.invoke mein chala jaayga ->like chain
    "language":"Python",
    "query":"Explain decoretor in simple word only in 20 words"
}))
```

```python
# Create chains -> FINAL WAY TO USE LLMs with langchain
chains = prompts | llm # bitewise or -> prompts ka output llm ke pass chala jaayga
chains
```

```python
res = chains.invoke({
    "language":"Python",
    "query":"Explain decoretor in simple word only in 20 words"
})
print(res.content[0]["text"])
```

```python
from langchain_core.output_parsers import StrOutputParser # it only provide us string from output
out = StrOutputParser() # object
chains = prompts | llm | out
res = chains.invoke({
    "language":"Python",
    "query":"Explain decoretor in simple word only in 20 words"
})
print(res) # no need to write res.content[0]["text"]
```

```python
def transform(res:str):
    return res.upper()
```

```python
from langchain_core.output_parsers import StrOutputParser # it only provide us string from output
out = StrOutputParser() # object
chains = prompts | llm | out | transform # now we will manippulate our outpur too
res = chains.invoke({
    "language":"Python",
    "query":"Explain decoretor in simple word only in 20 words"
})
print(res) # no need to write res.content[0]["text"]
```
